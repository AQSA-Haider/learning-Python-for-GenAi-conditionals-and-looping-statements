## Process multiple orders(for loop) 

```python
orders=[1200,2500,1800,1750,3000]
print(orders)
d=[]
fa=[]
```

```python
for order_amount in orders:
    tax=order_amount*(5/100)
    if order_amount>=2000:
        discount=order_amount*(15/100)
        subtotal=order_amount-discount
        final=subtotal+tax
        d.append(discount)
        fa.append(final)
        continue
    elif 1500<=order_amount>1000:
         discount=order_amount*(10/100)
         subtotal=order_amount-discount
         final=subtotal+tax
         d.append(discount)
         fa.append(final)
         continue
    elif 1000<=order_amount>1500:
         discount=order_amount*(7/100)
         subtotal=order_amount-discount
         final=subtotal+tax
         d.append(discount)
         fa.append(final)
         continue
    else:
         discount=order_amount*(0/100)
         subtotal=order_amount-discount
         final=subtotal+tax
    d.append(discount)
    fa.append(final)
        
print("Order Amount",orders)
print("Discount:",d)
print("Final Amount:",fa)
    
        
        
```

## Computing and printing the revenue

```python
for cost in orders:
    for discount in d:
        revenue= cost+(cost*(5/100))-discount
        break
        
print(revenue)
```

## User Menu(While loop + break/continue)

## Given orders list 

```python
orders=[1200,2500,1800,1750,3000]
print(orders)
d=[]
fa=[]
```

```python
for order_amount in orders:
    tax=order_amount*(5/100)
    if order_amount>=2000:
        discount=order_amount*(15/100)
        subtotal=order_amount-discount
        final=subtotal+tax
        d.append(discount)
        fa.append(final)
        continue
    elif 1500<=order_amount>1000:
         discount=order_amount*(10/100)
         subtotal=order_amount-discount
         final=subtotal+tax
         d.append(discount)
         fa.append(final)
         continue
    elif 1000<=order_amount>1500:
         discount=order_amount*(7/100)
         subtotal=order_amount-discount
         final=subtotal+tax
         d.append(discount)
         fa.append(final)
         continue
    else:
         discount=order_amount*(0/100)
         subtotal=order_amount-discount
         final=subtotal+tax
    d.append(discount)
    fa.append(final)
        
print("Order Amount",orders)
print("Discount:",d)
print("Final Amount:",fa)
```

## Creating a user action menu using simple while loop:
   asking the user to choose, until they press 'q' to Quit.
   # Menu options:
   :1--Add order_amount to running list(using(input)).
   :2--Shows all orders and totals after applying discounts.
   :q--Quit. 

```python
while True:
    print("Choose from:a)1 b)2 c)q")
    choose=input("choose action: (a).1  (b).2 (c).q")
    if choose == '1':
       print("enter amount of order you want to add:")
       amount=int(input("enter order amount you want to add"))
       orders.append(amount)
       tax=amount*(5/100)
       if amount>=2000:
          discount=amount*(15/100)
          subtotal=amount-discount
          final=subtotal+tax
          d.append(discount)
          fa.append(final)
          continue
       elif 1500<=amount>1000:
            discount=amount*(10/100)
            subtotal=amount-discount
            final=subtotal+tax
            d.append(discount)
            fa.append(final)
            continue
       elif 1000<=amount>1500:
            discount=amount*(7/100)
            subtotal=amount-discount
            final=subtotal+tax
            d.append(discount)
            fa.append(final)
            continue
       else:
           discount=amount*(0/100)
           subtotal=amount-discount
           final=subtotal+tax
       d.append(discount)
       fa.append(final)
         
    elif choose == '2':
         print("orders:",orders)
         print("Final Amount:",fa)
         
    elif choose =='q':
        print("Quit")
        break
    else:
        print("invalid selection!!")
        break
```

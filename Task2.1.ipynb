## Discount Rules(if/elif/else) 

## creating a variable(order_amount) that reads an integer using input() :
where the user will enter the amount of there order.

```python
order_amount=int(input("Enter the order amount of your product"))
n=order_amount
print(order_amount)
```

## different operations to be applied and variables to be used:

```python
d15=n*(15/100)
d10=n*(10/100)
d7=n*(7/100)
tax=n*(5/100)
subtotal=0
total=0
```

## Checking different conditions using if/elif/else:
#if order_amount>=2000:
 the discount will be 15%.
#if 1500<= order_amount<2000:
 the discount will be 10%.
#if 1000<= order_amount<1500:
 the discount will be 7%.
#if  order_amount<1000:
 the discount will be 0% thus,
## subtotal=order_amount-discount. 
#also, a fixed tax of 5% is applied on every order_amount
thus,
## final amount=subtotal + tax.




   

```python
if order_amount>=2000:
    subtotal=n-d15
    print("the discount applied will be 15%")
    print("the subtotal:",subtotal)
    print("fixed tax:",tax)
    print("final amount:",subtotal+tax)
elif 1500<= order_amount<2000:
    subtotal=n-d10
    print("the discount applied will be 10%")
    print("the subtotal:",subtotal)
    print("fixed tax:",tax)
    print("final amount:",subtotal+tax)
elif 1000<= order_amount<1500:
    subtotal=n-d7
    print("the discount applied will be 7%")
    print("the subtotal:",subtotal)
    print("fixed tax:",tax)
    print("final amount:",subtotal+tax)
else:
    subtotal=n
    print("the discount applied will be 0%")
    print("the subtotal:",subtotal)
    print("fixed tax:",tax)
    print("final amount:",subtotal+tax)
    

```

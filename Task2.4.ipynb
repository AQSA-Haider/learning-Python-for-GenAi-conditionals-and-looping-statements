## Loop Control With Conditions:

## Storing Given Daily Sales As a List(daily)

```python
daily=[200,50,0,400,50,-1,300]
print(daily)
```

## Applying the conditions with loop control statements :
#if a day sales is -1,it is treated as corrupted data and break the loop(stop processing).
#if a day sales is 0, it is treated as day with no sales and continue the loop(skip adding the revenue).
#valid sales has been added to total_sales nd the running total is being printed.

```python
total_sales=0
for day_sales in daily:
    if day_sales==-1:
        print("Corrupted Data Entered")
        break
    elif day_sales==0:
        print("it was a no sales day")
        continue
    else:
        total_sales=total_sales+day_sales
        print("running Total:",total_sales)
       
```

## Printing the final total sales

```python
print("Final Total Sales:",total_sales) 
```
